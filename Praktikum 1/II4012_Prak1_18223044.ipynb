## Rule-based Chatbot using NLTK
Sumber: https://www.analyticsvidhya.com/blog/2021/07/build-a-simple-chatbot-using-python-and-nltk/

```python
import nltk
from nltk.chat.util import Chat, reflections

indonesian_reflections = {
    "saya": "Anda",
    "aku": "kamu",
    "saya ingin": "Anda ingin",
    "milik saya": "milik Anda",
    "i": "you",
    "my": "your"
}

patterns = [
    [
        r".*(jenis|macam|apa saja).*layanan.*kemahasiswaan.*",
        [
            JAWABAN_JENIS_LAYANAN_KEMAHASISWAAN,
            f"Tentu, berikut informasinya: {JAWABAN_JENIS_LAYANAN_KEMAHASISWAAN}",
            "ITB menyediakan berbagai layanan, di antaranya: " + JAWABAN_JENIS_LAYANAN_KEMAHASISWAAN
        ]
    ],
    [
        r".*(berapa lama|batas waktu|maksimal).*(menghuni|tinggal|di).*asrama.*",
        [
            JAWABAN_LAMA_ASRAMA,
            "Berdasarkan aturan, " + JAWABAN_LAMA_ASRAMA
        ]
    ],
    [
        r"Kenapa (.*) harus tahu (.*)?",
        ["Karena %2 sangat penting bagi %1 sebagai bagian dari civitas akademika."]
    ],
    [
        r".*(kewajiban).*(penerima|pemegang).*beasiswa.*",
        [JAWABAN_KEWAJIBAN_PENERIMA_BEASISWA]
    ],
    [
        r".*(program|apa saja|bentuk|kegiatan).*pengembangan karakter.*",
        [JAWABAN_PENGEMBANGAN_KARAKTER]
    ],
    [
        r".*(kewajiban).*organisasi kemahasiswaan.*(pelaporan|laporan|kegiatan).*",
        [JAWABAN_KEWAJIBAN_ORGANISASI_PELAPORAN]
    ],
    [
        r".*(persyaratan|syarat).*mahasiswa berprestasi.*(pascasarjana).*",
        [JAWABAN_SYARAT_MAWAPRES_PASCASARJANA]
    ],
    [
        r".*(sumber|dari mana).*(pendanaan|dana|pembiayaan).*organisasi.*",
        [JAWABAN_SUMBER_PENDANAAN_ORGANISASI]
    ],
    [
        r".*(bentuk|apa).*layanan kesehatan.*mahasiswa.*",
        [JAWABAN_LAYANAN_KESEHATAN]
    ],
    [
        r".*(syarat|berapa).*(minimal|jumlah).*anggota.*ukm.*",
        [JAWABAN_MINIMAL_ANGGOTA_UKM]
    ],
    [
        r".*(kategori|apa saja|jenis|macam).*penghargaan ganesha.*",
        [JAWABAN_KATEGORI_PENGHARGAAN_GANESHA]
    ],
    [
        r"hi|hey|hello|halo",
        ["Halo! Ada yang bisa saya bantu terkait peraturan ITB?", "Hai! Silakan ajukan pertanyaan Anda."]
    ],
    [
        r"quit|exit|keluar",
        ["Sampai jumpa!", "Terima kasih telah bertanya."]
    ],
    [
        r"(.*)",
        [FALLBACK, "Maaf, bisa ulangi pertanyaannya?", "Saya kurang mengerti, coba gunakan kata kunci lain."]
    ]
]

# Inisialisasi chatbot
chatbot = Chat(patterns, indonesian_reflections)

def main():
    print("="*60)
    print("CHATBOT NLTK")
    print("="*60)

    response_times = []

    while True:
        user_input = input("You: ")
        response = chatbot.respond(user_input)
        print("Bot:", response)
        if user_input.lower() in ['quit', 'exit', 'keluar']:
          if response_times:
                print("\n" + "="*40)
                print("📊 RANGKUMAN WAKTU RESPON NLTK")
                print("="*40)
                print(f"Total Pertanyaan : {len(response_times)}")
                print(f"Waktu Tercepat   : {min(response_times):.4f} detik")
                print(f"Waktu Terlama    : {max(response_times):.4f} detik")
                print(f"Rata-rata Waktu  : {(sum(response_times)/len(response_times)):.4f} detik")
                print("="*40)
          break
        start_t = time.time()
        response = chatbot.respond(user_input)
        end_t = time.time()
        dur = end_t - start_t
        response_times.append(dur)

        print(f"[Waktu respon: {dur:.4f} detik]")

if __name__ == "__main__":
    main()
```

## Chatbot WITH LLM

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Load Model & Tokenizer
MODEL_ID = "Qwen/Qwen2.5-1.5B-Instruct"

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {device}")
print(f"Memuat model: {MODEL_ID} ...")

tok = AutoTokenizer.from_pretrained(MODEL_ID, use_fast=True)
llm_model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32,
    low_cpu_mem_usage=True,
)
llm_model.to(device)
llm_model.eval()
print("Model siap!")

# Fungsi Chat (LLM-only, tanpa konteks dokumen)
@torch.inference_mode()
def chat_once(user_msg: str, max_new_tokens=512) -> str:
    messages = [
        {
            "role": "system",
            "content": (
                "Anda adalah asisten yang menjawab pertanyaan tentang Peraturan Kemahasiswaan "
                "Institut Teknologi Bandung (ITB). Jawab dalam Bahasa Indonesia dengan singkat dan jelas."
            ),
        },
        {"role": "user", "content": user_msg},
    ]
    prompt = tok.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tok(prompt, return_tensors="pt").to(device)
    input_len = inputs["input_ids"].shape[1]

    out = llm_model.generate(
        **inputs,
        do_sample=False,
        max_new_tokens=max_new_tokens,
        pad_token_id=tok.eos_token_id,
        eos_token_id=tok.eos_token_id,
    )[0]

    gen = out[input_len:]
    reply = tok.decode(gen, skip_special_tokens=True).strip()
    return reply

# 10 pertanyaan dari daftar pertanyaan yang sudah ditentukan
listQ = [
    "Apa saja jenis layanan kemahasiswaan yang disediakan oleh ITB?",
    "Berapa lama batas waktu maksimal mahasiswa dapat menghuni Asrama Mahasiswa ITB?",
    "Apa kewajiban utama dari seorang penerima beasiswa di ITB?",
    "Apa saja program Pengembangan Karakter mahasiswa ITB?",
    "Apa kewajiban Organisasi Kemahasiswaan ITB terkait pelaporan kegiatannya?",
    "Apa persyaratan umum bagi mahasiswa tingkat Pascasarjana untuk terpilih sebagai Mahasiswa Berprestasi?",
    "Dari mana saja sumber pendanaan organisasi kemahasiswaan di ITB?",
    "Apa bentuk layanan kesehatan bagi mahasiswa ITB?",
    "Berapa syarat minimal jumlah anggota untuk mendirikan UKM?",
    "Apa saja kategori Penghargaan Ganesha di ITB?",
]

# Main: untuk memilih mode
print("=" * 60)
print("CHATBOT LLM-ONLY (Tanpa Dokumen / Knowledge Base)")
print(f"Model: {MODEL_ID}")
print("=" * 60)
print("\nCatatan: Model menjawab MURNI dari pengetahuan internal (pre-training).")

print("Pilih mode:")
print("  [1] Jalankan otomatis 10 pertanyaan dari daftar pertanyaan yang telah ditentukan")
print("  [2] Chat manual dengan ChatBot")

mode = input("\nPilih mode (1/2): ").strip()

# Mode 1: Test otomatis
if mode == "1":
    print("\n" + "=" * 60)
    print("TEST OTOMATIS — 10 Pertanyaan")
    print("=" * 60)
    for i, q in enumerate(listQ, 1):
        print(f"\n[Q{i}] User: {q}")
        ans = chat_once(q)
        print(f"      Bot : {ans}")
        print("-" * 60)

# Mode 2: Chat manual
elif mode == "2":
    print("\n" + "=" * 60)
    print("CHAT MANUAL — Ketik pertanyaan Anda (ketik 'exit' untuk keluar)")
    print("=" * 60)

    response_times = []

    while True:
        try:
            q = input("\nUser: ").strip()
        except EOFError:
            break
        if not q or q.lower() == "exit":
            print("Terima kasih! Sesi LLM-only selesai.")
            if response_times:
                print("\n" + "="*40)
                print("📊 RANGKUMAN WAKTU RESPON LLM-ONLY")
                print("="*40)
                print(f"Total Pertanyaan : {len(response_times)}")
                print(f"Waktu Tercepat   : {min(response_times):.4f} detik")
                print(f"Waktu Terlama    : {max(response_times):.4f} detik")
                print(f"Rata-rata Waktu  : {(sum(response_times)/len(response_times)):.4f} detik")
                print("="*40)
            break
        start_t = time.time()
        ans = chat_once(q)
        end_t = time.time()
        dur = end_t - start_t
        response_times.append(dur)

        print(f"Bot : {ans}")
        print(f"[Waktu respon: {dur:.4f} detik]")
        print("-" * 60)

else:
    print("Mode tidak dikenali. Silakan jalankan ulang dan pilih 1 atau 2.")
```

# Chatbot RAG -LLM

Chatbot ini menggunakan dokumen PDF cerita fiksi “Sepuluh Sahabat ke Kota Senja”.
Secara garis besar, alurnya sebagai berikut:

1. Parsing PDF  → Ekstraksi teks halaman memakai pypdf.PdfReader. Teks dibersihkan (ligature, soft-hyphen, pemotongan “-\n”, spasi berlebih). Dokumen kemudian dipecah berdasarkan heading bernomor (jika ada); bila tidak terdeteksi, dipotong per ukuran tetap (default chunk_size=1200, overlap=200).


2. Struktur & Booster Domain Cerita → Konten diperkaya dengan booster pencarian untuk kode & istilah cerita:
C01–C10 (karakter), L01–L10 (lokasi), S1–S8 (adegan), K1–K4 (konflik), T01/T02 (tiket), R01/R02 (nota), P01 (peta), nama tokoh/lokasi (Aksa, Bela, Museum Riwayat Batu, Alun-Alun, Gedung Pusaka Angin, Danau Cermin, Pasar Malam Senja, Rumah Makan Panyawara, Terminal Caraka), serta jam penting 07:30, 09:00, 13:00. Booster ini meningkatkan skor kandidat yang relevan saat retrieve.

3. Embedding & Indexing (RAG) → Potongan teks diubah menjadi vektor menggunakan SentenceTransformer (BAAI/bge-m3) dengan normalisasi, lalu diindeks ke FAISS (IndexFlatIP) untuk semantic search. Parameter utama: k=12 untuk pencarian awal (top-k).

4. Re-ranking (opsional) → Hasil retrieve awal dapat diurut ulang oleh CrossEncoder (BAAI/bge-reranker-v2-m3) agar konteks paling relevan naik ke atas (default topk_final=6). Fitur ini bisa dimatikan lewat konfigurasi jika RAM terbatas.

5. LLM Generation → Jawaban disintesis oleh Qwen/Qwen2.5-1.5B-Instruct. Prompt dibentuk di fungsi make_messages, menegaskan aturan: jawab singkat, jelas, dan HANYA dari konteks. Jika tidak ada informasi relevan, bot menjawab: “Tidak ditemukan dalam naskah.” (tanpa menyertakan rujukan).

Fungsi kunci:

build_docs() → baca PDF, bersihkan teks, pecah jadi potongan.
retrieve() → FAISS search + booster kode/istilah cerita (+ opsional re-ranking).
make_messages() → susun prompt berisi konteks terpilih.
generate_answer() → panggil LLM untuk menghasilkan jawaban.
chat(query) → pipeline ujung-ke-ujung (retrieve → generate).

```python
import os, re
from pypdf import PdfReader

# Ganti ke path PDF kamu di Colab
PDF_PATH = "/content/Peraturan_Kemahasiswaan_2022.pdf"

def verify_pdf(path: str) -> bool:
    if not os.path.isfile(path):
        raise FileNotFoundError(f"Tidak menemukan PDF di: {path}")
    try:
        reader = PdfReader(path)
        n_pages = len(reader.pages)
        if n_pages == 0:
            raise ValueError("PDF tidak memiliki halaman.")
        print(f"PDF OK — jumlah halaman: {n_pages}")
        return True
    except Exception as e:
        raise RuntimeError(f"Gagal membaca PDF: {e}")

def extract_pages(path: str):
    """Ekstrak teks per halaman (tanpa OCR). Halaman kosong di-skip."""
    reader = PdfReader(path)
    pages = []
    for i, page in enumerate(reader.pages, start=1):
        txt = page.extract_text() or ""
        # Normalisasi ringan
        txt = txt.replace("\u00ad", "")     # soft hyphen
        txt = txt.replace("-\n", "")        # sambung kata yang terpotong
        txt = re.sub(r"\s+", " ", txt).strip()
        if txt:
            pages.append({"page": i, "text": txt})
    print(f"Halaman dengan teks: {len(pages)}")
    return pages

# Verifikasi & contoh ekstraksi
# verify_pdf(PDF_PATH) # This will fail for URL, will need to be downloaded first
# pages_text = extract_pages(PDF_PATH)
# if pages_text:
#     print("Cuplikan halaman 1 (200 karakter):")
#     print(pages_text[0]["text"][:200])
```

```python
# Cell 4 - PDF processing (tanpa OCR)

def clean_text(s: str) -> str:
    """Bersihkan teks dari karakter aneh & normalisasi spasi/baris."""
    s = (s or "")
    s = s.replace("\x00", " ").replace("\ufb01", "fi").replace("\ufb02", "fl")  # ligatures umum
    s = s.replace("\u00ad", "")        # soft hyphen
    s = s.replace("-\n", "")           # sambung kata terpotong di line break
    s = re.sub(r"\n{3,}", "\n\n", s)   # kompres newline berlebih
    s = re.sub(r"[ \t]+", " ", s)      # kompres spasi/tab
    return s.strip()

def read_pdf_text(path: str, min_chars: int = 50) -> Tuple[str, List[Tuple[int, str]]]:
    print(f"Membaca PDF: {os.path.basename(path)}")

    reader = PdfReader(path)
    pages: List[Tuple[int, str]] = []

    for i, p in enumerate(reader.pages, start=1):
        raw = p.extract_text() or ""
        t = clean_text(raw)
        pages.append((i, t))
        print(f"  Halaman {i}: {len(t)} karakter")

    # Peringatan untuk halaman yang tampak kosong/scan
    suspect = [i for (i, t) in pages if len(t) < min_chars]
    if suspect:
        print(f"Peringatan: {len(suspect)} halaman terdeteksi minim teks {suspect}. "
              f"Kemungkinan berbasis gambar. Mode ini PDF-only, OCR dinonaktifkan.")

    full = "\n\n".join([f"[page {i}]\n{t}" for i, t in pages])
    total_chars = sum(len(t) for _, t in pages)
    print(f"Total: {len(pages)} halaman, {total_chars} karakter (tanpa OCR)")

    return full, pages

# Contoh pakai (opsional):
full_text, pages_list = read_pdf_text(PDF_PATH)
print(full_text[:500])
```

```python
import re, os

all_docs = []

for path in PDFS:
    print(f"\n=== Proses: {os.path.basename(path)} ===")
    full_text, page_texts = read_pdf_text(path)
    source_name = os.path.basename(path)

    # Gunakan chunking sederhana (1000 karakter) karena ini dokumen hukum/peraturan
    step = 1000
    overlap = 150
    for i in range(0, len(full_text), step - overlap):
        chunk = full_text[i:i+step].strip()
        if not chunk: continue

        # Cari perkiraan halaman
        page_hint = None
        for pg, txt in page_texts:
            if chunk[:50] in txt:
                page_hint = pg
                break

        all_docs.append({
            "text": chunk,
            "section": "Peraturan Kemahasiswaan",
            "source": source_name,
            "page_hint": page_hint
        })

print(f"Total chunks generated: {len(all_docs)}")
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np
import faiss

if not all_docs:
    print("Warning: all_docs is empty. Running fallback read...")
    full_text, _ = read_pdf_text(PDF_PATH)
    all_docs = [{"text": full_text[i:i+1000], "section": "Manual", "source": "PDF", "page_hint": None} for i in range(0, len(full_text), 900)]

EMB_MODEL = "BAAI/bge-m3"
print(f"Memuat embedding model: {EMB_MODEL}")
embedder = SentenceTransformer(EMB_MODEL)

corpus = [d["text"] for d in all_docs]
X = embedder.encode(corpus, batch_size=32, normalize_embeddings=True, show_progress_bar=True)
X = np.array(X, dtype="float32")

index = faiss.IndexFlatIP(X.shape[1])
index.add(X)
print(f"Index FAISS siap dengan {index.ntotal} vektor.")
```

```python
# Cell 8 - Setup reranker (opsional tapi disarankan)
from sentence_transformers import CrossEncoder
import numpy as np

RERANKER = "BAAI/bge-reranker-v2-m3"
print(f"Memuat reranker: {RERANKER}")
reranker = CrossEncoder(RERANKER)  # otomatis pakai GPU jika tersedia (via torch)

def rerank_pairs(query: str, docs: List[Dict], topk: int = 5) -> List[Dict]:
    """
    Re-ranking doc hasil retrieval awal (mis. dari FAISS) dengan CrossEncoder.
    Mengembalikan hingga topk dokumen terbaik, menambahkan 'rerank_score'.
    """
    if not docs:
        return []
    k = min(topk, len(docs))
    pairs = [[query, d["text"]] for d in docs]
    scores = np.asarray(reranker.predict(pairs), dtype="float32")
    order = np.argsort(-scores)[:k]
    return [docs[i] | {"rerank_score": float(scores[i])} for i in order]
```

```python
# Cell 9 - Setup LLM
LLM = "Qwen/Qwen2.5-1.5B-Instruct"

device = "cuda" if torch.cuda.is_available() else "cpu"
tokenizer = AutoTokenizer.from_pretrained(LLM, use_fast=True)
model = AutoModelForCausalLM.from_pretrained(
    LLM,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32,
    low_cpu_mem_usage=True,
)
model.to(device)
model.eval()

def make_messages(query: str, ctxs: List[Dict]) -> List[Dict]:
    bullets = []
    for c in ctxs:
        snippet = re.sub(r"\s{2,}", " ", c["text"]).strip()[:900]
        bullets.append(f"- {snippet}")
    context_block = "\n".join(bullets) if bullets else "(konteks kosong)"
    return [
        {
            "role": "system",
            "content": (
                "Anda asisten akademik yang menjawab berdasarkan konteks yang diberikan. "
                "Jawab SINGKAT, JELAS, dan HANYA dari konteks. "
                "Jika tidak ada informasi relevan, balas: 'Tidak ditemukan dalam dokumen.' "
                "Jangan sertakan rujukan/sumber dalam jawaban."
            ),
        },
        {
            "role": "user",
            "content": (
                f"KONTEKS:\n{context_block}\n\n"
                f"PERTANYAAN:\n{query}\n\n"
                f"Jawab singkat dan jelas tanpa menyertakan rujukan."
            ),
        },
    ]

@torch.inference_mode()
def generate_answer(messages, max_new_tokens=256):
    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    input_len = inputs["input_ids"].shape[1]
    out = model.generate(
        **inputs,
        do_sample=False,
        max_new_tokens=max_new_tokens,
        pad_token_id=tokenizer.eos_token_id,
        eos_token_id=tokenizer.eos_token_id,
    )[0]
    gen = out[input_len:]
    txt = tokenizer.decode(gen, skip_special_tokens=True).strip()
    return re.sub(r"^assistant\s*:\s*", "", txt, flags=re.I) or "Tidak ditemukan dalam dokumen."
```

```python
def retrieve(query: str, k: int = 5) -> List[Dict]:
    qv = np.array(embedder.encode([query], normalize_embeddings=True), dtype="float32")
    D, I = index.search(qv, k)

    results = []
    for idx in I[0]:
        if idx != -1:
            results.append(all_docs[idx])
    return results

@torch.inference_mode()
def chat(query: str) -> Dict:
    # Logika dasar untuk sapaan atau identitas
    q_lower = query.lower()
    if any(greet in q_lower for greet in ["halo", "hi ", "hello", "siapa anda", "siapa kamu"]):
        if "siapa" in q_lower:
            return {"answer": "Saya adalah Asisten AI yang siap membantu Anda menjawab pertanyaan seputar Peraturan Kemahasiswaan ITB.", "contexts": []}
        return {"answer": "Halo! Ada yang bisa saya bantu terkait peraturan kemahasiswaan ITB?", "contexts": []}

    ctxs = retrieve(query)
    context_text = "\n".join([f"- {c['text']}" for c in ctxs])

    messages = [
        {
            "role": "system",
            "content": (
                "Anda adalah Asisten Akademik AI yang membantu menjelaskan Peraturan Kemahasiswaan ITB. "
                "Gunakan konteks yang diberikan untuk menjawab. Jawablah dengan ramah namun profesional. "
                "Jika informasi tidak ada di konteks, katakan bahwa Anda tidak menemukan informasinya di dokumen peraturan."
            )
        },
        {
            "role": "user",
            "content": f"KONTEKS DOKUMEN:\n{context_text}\n\nPERTANYAAN: {query}"
        }
    ]

    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    out = model.generate(
        **inputs,
        max_new_tokens=300,
        do_sample=True, # Mengaktifkan sampling agar jawaban tidak terlalu kaku
        temperature=0.7,
        top_p=0.9,
        pad_token_id=tokenizer.eos_token_id
    )

    ans = tokenizer.decode(out[0][inputs['input_ids'].shape[1]:], skip_special_tokens=True).strip()
    return {"answer": ans, "contexts": ctxs}

print("RAG Chatbot is now smarter and friendlier!")
```
